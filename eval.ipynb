```python
# Merge label results

f = open('train.jsonl','r')
l = f.readlines()
res =[ ]
for l_ in l:
    res.append(json.loads(l_))
```

```python
gt_data = pd.DataFrame(res)
```

```python
gt_data
```

```python
merged_df = pd.merge(gt_data, merged_df, on='claim')
merged_df
```

```python
from sklearn.metrics import classification_report

clean_df = merged_df.dropna(subset=['classification'])
report = classification_report(clean_df['label'].to_list(), clean_df['classification'].to_list())

print(report)
```
